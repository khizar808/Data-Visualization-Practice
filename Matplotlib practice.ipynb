# Matplotlib Library

# Line plot:

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.arange(1,11)
x
```

```python
y = x*2
y
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.plot(x,y,color='r',linestyle=':',linewidth=2)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x
```

```python
y1 = 2*x
y1
```

```python
y2 = 3*x
y2
```

```python
plt.plot(x,y1,color='r',linestyle=':')
plt.plot(x,y2,color='g',linestyle='-')
plt.grid(True)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y1,color='r',linestyle=':')

plt.subplot(1,2,2)
plt.plot(x,y2,color='g',linestyle='-')

plt.show()
```

# Bar plot:

```python
student = {'Usama':80,'Khizar':60,'Talha':70,'Hamza':90}
```

```python
names = list(student.keys())
```

```python
names
```

```python
values = list(student.values())
```

```python
values
```

```python
plt.bar(names,values)
plt.title("Result of Students")
plt.xlabel("Names")
plt.ylabel("Obtained Marks")
plt.show()
```

```python
plt.barh(names,values,color='r')
plt.title("Result of Students")
plt.xlabel("Names")
plt.ylabel("Obtained Marks")
plt.grid(True)
plt.show()
```

# Scatter plot:

```python
x = [2,3,6,8,10,5,9,8]
y = [10,5,7,9,1,2,3,4]
```

```python
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.scatter(x,y,color='r',marker='*')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Adding two markers or points in one scatter plot
```

```python
x = [2,3,6,8,10,5,9,8]
a = [10,5,7,9,1,2,3,4]
b = [4,6,15,12,14,3,4,3]
```

```python
plt.scatter(x,a,color='r',marker='*',s=100)
plt.scatter(x,b,color='k',marker='.',s=150)

plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

```python
# Making subplots
```

```python
plt.subplot(2,1,1)
plt.scatter(x,a,color='r',marker='*',s=100)

plt.subplot(2,1,2)
plt.scatter(x,b,color='k',marker='.',s=150)

plt.show()
```

# Box plot:

It is also called five number summary.
1) minimum
2) 1st quartile means 25% of data in the dataset
3) 2nd quartile means 50% of data in the dataset
4) 3rd quartile means 75% of data in the dataset
5) maximum

```python
one = [1,2,3,4,5]
two = [5,6,7,8,9]
three = [3,4,5,6,7]
data = list([one,two,three])
```

```python
data
```

```python
plt.boxplot(data)
plt.show()
```

# Violin plot:

```python
one = [1,2,3,4,5]
two = [5,6,7,8,9]
three = [3,4,5,6,7]
data = list([one,two,three])
plt.violinplot(data,showmedians=True)
plt.show()
```

# Pie chart:

```python
fruits = ['Mango','Grapes','Orange','Apple']
quantity = [50,20,70,100]
```

```python
plt.pie(quantity,labels=fruits)
plt.show()
```

```python
# Adding percentage to pie chart or pie plot
plt.pie(quantity,labels=fruits,autopct='%0.1f%%')
plt.show()
```

```python
plt.pie(quantity,labels=fruits,autopct='%0.1f%%',colors=['yellow','black','green','orange'])
plt.show()
```

```python
plt.pie(quantity,labels=fruits,autopct='%0.1f%%',colors=['y','b','g','r'])
plt.show()
```
